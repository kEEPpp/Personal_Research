```python
def aakr(X, sigma=1.0):
    """
    Auto-Associative Kernel Regression (AAKR) 수행
    :param X: 입력 데이터 (n_samples, n_features)
    :param sigma: 커널 대역폭
    :return: 복원된 데이터, 복원 오차
    """
    K = rbf_kernel(X, X, gamma=1/(2*sigma**2))  # RBF Kernel
    weights = K / K.sum(axis=1, keepdims=True)  # 가중치 정규화
    X_hat = np.dot(weights, X)  # 복원 데이터
    
    reconstruction_error = np.linalg.norm(X - X_hat, axis=1)  # 복원 오차 계산
    return X_hat, reconstruction_error
```

```python
X_hat, reconstruction_error = aakr(data, sigma=1.0)

# 이상값 탐지 (Threshold 설정)
threshold = np.percentile(reconstruction_error, 95)  # 상위 5%를 이상값으로 설정
predicted_anomalies = reconstruction_error > threshold

# 결과 시각화
plt.scatter(data[:, 0], data[:, 1], c=predicted_anomalies, cmap="coolwarm", label="Data Points")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title("Anomaly Detection using AAKR")
plt.legend()
plt.colorbar(label="Anomaly Score")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics.pairwise import rbf_kernel

# 데이터 생성
np.random.seed(42)
normal_data = np.random.normal(0, 1, (100, 2))  # 정상 데이터
anomalous_data = np.random.normal(5, 1, (10, 2))  # 이상 데이터
data = np.vstack([normal_data, anomalous_data])  # 전체 데이터
labels = np.array([0] * 100 + [1] * 10)  # 라벨 (정상=0, 이상=1)

# AAKR 알고리즘 구현
def aakr(X, sigma=1.0):
    K = rbf_kernel(X, X, gamma=1/(2*sigma**2))  # RBF Kernel 계산
    weights = K / K.sum(axis=1, keepdims=True)  # 가중치 정규화
    X_hat = np.dot(weights, X)  # 데이터 재구성
    reconstruction_error = np.linalg.norm(X - X_hat, axis=1)  # 복원 오차 계산
    return X_hat, reconstruction_error

# 실행 및 이상 탐지
X_hat, reconstruction_error = aakr(data, sigma=1.0)
threshold = np.percentile(reconstruction_error, 95)  # 상위 5%를 이상값으로 설정
predicted_anomalies = reconstruction_error > threshold

# 결과 시각화
plt.scatter(data[:, 0], data[:, 1], c=predicted_anomalies, cmap="coolwarm", label="Data Points")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title("Anomaly Detection using AAKR")
plt.legend()
plt.colorbar(label="Anomaly Score")
plt.show()
```
