# threshold based peak detection

```python
window_len = 5
image_col = 5
fig = None

for idx, (wid, wafer) in enumerate(data.groupby('WAFER_ID')):
    if idx % image_col == 0:
        if fig:
            plt.show()  # 이전 figure를 보여줌
        fig, ax = plt.subplots(1, image_col, figsize=(20, 2))

    col_idx = idx % image_col
    raw_wafer = wafer['PARAMETER_VALUE'].iloc[window_len - 1:].values
    roll_wafer = wafer['PARAMETER_VALUE'].rolling(window_len).median().iloc[window_len - 1:].values
    error = raw_wafer - roll_wafer

    error_thresholds = [
        np.quantile(error, 0.95),
        3 * np.std(error, ddof=1) + np.mean(error),
        #np.max(error) * 0.8
        #(np.quantile(error, 0.75) - np.quantile(error, 0.25)) - (np.max(error) - np.min(error))
        #(np.max(error) - np.min(error)) / np.mean(error)
    ]

    ax[col_idx].plot(error)
    for threshold, color, label in zip(error_thresholds, ['red', 'blue', 'green'], ['thres1', 'thres2', 'thres3']):
        ax[col_idx].axhline(threshold, color=color, linestyle='--', alpha=0.5, label=label)
    ax[col_idx].legend()
    ax[col_idx].set_title(f'Wafer ID: {wid}')

# 마지막 figure를 보여줌
if fig:
    plt.show()
```

# PCA peak detection

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from scipy.signal import find_peaks

# 데이터 생성 (샘플 데이터)
# np.random.seed(42)
# data = np.sin(np.linspace(0, 20, 500)) + np.random.normal(0, 0.1, 500)
# data[100:110] += 2  # 인위적인 피크 추가
# data[300:305] -= 2
t = np.linspace(0, 1, data_reshaped[0].shape[0])
# PCA를 사용한 재구성 오류 계산
def compute_reconstruction_error(data, n_components=1):
    # 데이터를 2D 형태로 변환 (PCA 입력 형식)
    #data_reshaped = data.reshape(-1, 1)
    data_reshaped = data

    # PCA 적용
    pca = PCA(n_components=n_components)
    pca.fit(data_reshaped)
    
    # 데이터 재구성
    reconstructed = pca.inverse_transform(pca.transform(data_reshaped))

    # 재구성 오류 계산
    #return data_reshaped, reconstructed
    reconstruction_error = np.abs(data_reshaped - reconstructed)
    return reconstruction_error

# 재구성 오류 계산
reconstruction_error = compute_reconstruction_error(data_reshaped)

# 피크 감지 (scipy의 find_peaks 사용)
#threshold = np.percentile(reconstruction_error, 99)  # 상위 5%를 기준으로 설정
#peaks, _ = find_peaks(reconstruction_error, height=threshold)

fig = None

for idx, (wafer, error) in enumerate(zip(data_reshaped, reconstruction_error)):
    if idx % image_col == 0:
        if fig:
            plt.tight_layout()
            plt.show()
        fig, ax = plt.subplots(2, image_col, figsize=(20,5))

    col_idx = idx % image_col

    ax[0, col_idx].set_title(f'wafer ID: wafer{idx+1}')
    ax[1, col_idx].set_title(f'reconstruction error')
    ax[0, col_idx].plot(t, wafer)
    ax[1, col_idx].plot(t, error)

if fig:
    plt.tight_layout()
    plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from skfda import FDataGrid
from skfda.preprocessing.dim_reduction.feature_extraction import FPCA
from scipy.signal import find_peaks

# 데이터 생성 (여러 샘플로 확장)
# np.random.seed(42)
t = np.linspace(0, 1, data_reshaped[0].shape[0])
# data = np.vstack([
#     np.sin(2 * np.pi * 5 * t) + np.random.normal(0, 0.1, 500),
#     np.sin(2 * np.pi * 7 * t) + np.random.normal(0, 0.1, 500)
# ])

image_col = 5
# FDataGrid로 변환
fd = FDataGrid(data_matrix=data_reshaped, grid_points=t)

# FPCA를 사용한 재구성 오류 계산
def compute_fpca_reconstruction_error(fd, n_components=2):
    fpca = FPCA(n_components=n_components)
    fpca.fit(fd)
    reconstructed_fd = fpca.inverse_transform(fpca.transform(fd))
    reconstruction_error = np.abs(fd.data_matrix - reconstructed_fd.data_matrix)  # 각 샘플별로 오류 계산
    return reconstruction_error

# 재구성 오류 계산
reconstruction_error = compute_fpca_reconstruction_error(fd)

# 첫 번째 샘플에 대해 피크 감지
error_sample_0 = reconstruction_error[0]  # 첫 번째 샘플의 오류
threshold = np.percentile(error_sample_0, 95)  # 상위 5%를 기준으로 설정
peaks, _ = find_peaks(error_sample_0.flatten(), height=threshold)


fig = None

for idx, (wafer, error) in enumerate(zip(data_reshaped, reconstruction_error)):
    if idx % image_col == 0:
        if fig:
            plt.tight_layout()
            plt.show()
        fig, ax = plt.subplots(2, image_col, figsize=(20,5))

    col_idx = idx % image_col

    ax[0, col_idx].set_title(f'wafer ID: wafer{idx+1}')
    ax[1, col_idx].set_title(f'reconstruction error')
    ax[0, col_idx].plot(t, wafer)
    ax[1, col_idx].plot(t, error)

if fig:
    plt.tight_layout()
    plt.show()
# # 결과 시각화
# plt.figure(figsize=(12, 6))
# plt.subplot(2, 1, 1)
# plt.plot(t, data[0], label="Original Data (Sample 1)")
# #plt.plot(t, data[1])
# plt.scatter(t[peaks], data[0][peaks], color='red', label="Detected Peaks")  # 샘플 1의 피크
# plt.legend()
# plt.title("Original Data with Detected Peaks")

# plt.subplot(2, 1, 2)
# plt.plot(t, error_sample_0, label="Reconstruction Error (Sample 1)")
# plt.axhline(threshold, color='red', linestyle='--', label="Threshold")
# plt.legend()
# plt.title("Reconstruction Error (Sample 1)")

# plt.tight_layout()
# plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal

def mexican_hat_wavelet(t, sigma):
    return (1 / (np.sqrt(2 * np.pi) * sigma)) * (1 - (t**2 / sigma**2)) * np.exp(-(t**2 / (2 * sigma**2)))

def wavelet_transform(data, scales, wavelet):
    n = len(data)
    out = np.zeros((len(scales), n))
    for i, scale in enumerate(scales):
        wavelet_data = wavelet(np.arange(n) - n // 2, scale)
        out[i, :] = np.convolve(data, wavelet_data, mode='same')
    return out

# 시계열 데이터 생성
# t = np.linspace(0, 10, 1000)
# data = np.sin(t) + 0.5 * np.sin(5 * t) + 0.1 * np.random.randn(len(t))

t = np.linspace(0, 1, data_reshaped[0].shape[0])

# 웨이블릿 변환 적용
scales = np.arange(1, 31, 1)

cwt_list = []
for d in data_reshaped:
    cwt = wavelet_transform(d, scales, mexican_hat_wavelet)
    cwt_list.append(cwt)
    

fig = None

for idx, wafer in enumerate(wafer_list):
    if idx % image_col == 0:
        if fig:
            plt.tight_layout()
            #plt.legend()
            plt.show()
        fig, ax = plt.subplots(2, image_col, figsize=(20,5))

    col_idx = idx % image_col

    raw_wafer = wafer['PARAMETER_VALUE'].iloc[window_len - 1:].values
    roll_wafer = wafer['PARAMETER_VALUE'].rolling(window_len).median().iloc[window_len - 1:].values
    error = raw_wafer - roll_wafer
    
    cwt = wavelet_transform(error, scales, mexican_hat_wavelet)    
    peaks = signal.find_peaks(cwt[15])[0]  # 특정 스케일에서 피크 감지

    t = np.linspace(0, 1, len(error))
    ax[0, col_idx].set_title(f'wafer ID: wafer{idx+1}')
    ax[1, col_idx].set_title(f'wafer_coeff')
    
    #ax[1, col_idx].colorbar(label='wavelet coeff')
    
    ax[0, col_idx].plot(t, error)
    ax[0, col_idx].plot(t[peaks], error[peaks], 'ro', label='detected peak')
    ax[1, col_idx].imshow(cwt, aspect='auto', extent=[0, 10, 1, 30], cmap='jet', interpolation='nearest')

if fig:
    #plt.legend()
    plt.tight_layout()
    plt.show()

# 시각화
# plt.figure(figsize=(12, 8))

# plt.subplot(211)
# plt.plot(t, data)
# plt.title('raw timeseries data')
# plt.xlabel('time')
# plt.ylabel('amplitude')

# plt.subplot(212)
# plt.imshow(cwt, aspect='auto', extent=[0, 10, 1, 30], cmap='jet', interpolation='nearest')
# plt.colorbar(label='wavelet coeff')
# plt.title('Mexican Hat Wavelet transformation')
# plt.xlabel('time')
# plt.ylabel('scale')

# plt.tight_layout()
# plt.show()

# 피크 감지
# peaks = signal.find_peaks(cwt[15])[0]  # 특정 스케일에서 피크 감지

# plt.figure(figsize=(12, 4))
# plt.plot(t, data, label='raw data')
# plt.plot(t[peaks], data[peaks], 'ro', label='detected peak')
# plt.title('peak detection using wavelet')
# plt.xlabel('time')
# plt.ylabel('amplitude')
# plt.legend()
# plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from scipy.signal import find_peaks
from sklearn.ensemble import IsolationForest
from sklearn.cluster import DBSCAN
from statsmodels.tsa.seasonal import seasonal_decompose
from scipy.stats import zscore
from sklearn.svm import OneClassSVM
import pywt

# 1. Threshold-Based Detection
# 간단한 상/하한선 기반 탐지

def threshold_based_detection(data, threshold):
    peaks = np.where(data > threshold)[0]
    return peaks

# 2. PCA Reconstruction Error
# PCA를 사용한 재구성 오류 기반 탐지

def pca_reconstruction_detection(data, n_components=1):
    pca = PCA(n_components=n_components)
    transformed = pca.fit_transform(data.reshape(-1, 1))
    reconstructed = pca.inverse_transform(transformed)
    reconstruction_error = np.abs(data - reconstructed.flatten())
    peaks, _ = find_peaks(reconstruction_error, height=np.mean(reconstruction_error) + 3 * np.std(reconstruction_error))
    return peaks

# 3. Isolation Forest
# 이상치 탐지 모델로 탐지

def isolation_forest_detection(data):
    model = IsolationForest(contamination=0.05, random_state=42)
    preds = model.fit_predict(data.reshape(-1, 1))
    peaks = np.where(preds == -1)[0]
    return peaks

# 4. DBSCAN
# 밀도 기반 이상치 탐지

def dbscan_detection(data, eps=0.5, min_samples=5):
    db = DBSCAN(eps=eps, min_samples=min_samples)
    labels = db.fit_predict(data.reshape(-1, 1))
    peaks = np.where(labels == -1)[0]
    return peaks

# 5. Seasonal Decomposition
# 계절성 분해를 사용한 이상치 탐지

def seasonal_decompose_detection(data, period):
    result = seasonal_decompose(data, period=period, model='additive', extrapolate_trend='freq')
    residual = result.resid
    peaks = np.where(np.abs(zscore(residual)) > 3)[0]
    return peaks

# 6. Z-Score 기반 탐지
# Z-Score를 활용한 탐지

def zscore_detection(data):
    z_scores = zscore(data)
    peaks = np.where(np.abs(z_scores) > 3)[0]
    return peaks

# 7. Wavelet Transform 기반 탐지
# 웨이블릿 변환으로 이상치 탐지

def wavelet_transform_detection(data, wavelet='db4', level=1):
    coeff = pywt.wavedec(data, wavelet, level=level)
    coeff[-1] = np.zeros_like(coeff[-1])
    reconstructed = pywt.waverec(coeff, wavelet)
    peaks, _ = find_peaks(np.abs(data - reconstructed), height=np.mean(data - reconstructed) + 3 * np.std(data - reconstructed))
    return peaks

# 8. One-Class SVM
# SVM을 사용한 이상치 탐지

def one_class_svm_detection(data, nu=0.05):
    model = OneClassSVM(nu=nu, kernel='rbf', gamma=0.1)
    model.fit(data.reshape(-1, 1))
    preds = model.predict(data.reshape(-1, 1))
    peaks = np.where(preds == -1)[0]
    return peaks

# 9. Moving Average + Threshold
# 이동 평균과 임계값을 사용한 탐지

def moving_average_detection(data, window_size=5, threshold=2):
    rolling_mean = pd.Series(data).rolling(window=window_size).mean()
    diff = np.abs(data - rolling_mean)
    peaks = np.where(diff > threshold * np.std(data))[0]
    return peaks

# 10. Dynamic Time Warping (DTW)
# DTW 기반 이상치 탐지
from dtaidistance import dtw

def dtw_detection(data, reference):
    distance = dtw.distance(data, reference)
    peaks = np.where(distance > np.mean(distance) + 3 * np.std(distance))[0]
    return peaks

# 데이터 예제 및 실행
np.random.seed(42)
data = np.sin(np.linspace(0, 20, 1000)) + np.random.normal(0, 0.1, 1000)
data[500:510] += 3  # 인위적 피크

# 각 방법 실행
threshold_peaks = threshold_based_detection(data, threshold=1.5)
pca_peaks = pca_reconstruction_detection(data)
isolation_peaks = isolation_forest_detection(data)
dbscan_peaks = dbscan_detection(data)
seasonal_peaks = seasonal_decompose_detection(data, period=50)
zscore_peaks = zscore_detection(data)
wavelet_peaks = wavelet_transform_detection(data)
svm_peaks = one_class_svm_detection(data)
moving_avg_peaks = moving_average_detection(data)

# 결과 시각화
plt.figure(figsize=(12, 6))
plt.plot(data, label='Data')
plt.scatter(threshold_peaks, data[threshold_peaks], color='red', label='Threshold Peaks')
plt.scatter(pca_peaks, data[pca_peaks], color='green', label='PCA Peaks')
plt.scatter(isolation_peaks, data[isolation_peaks], color='blue', label='Isolation Peaks')
plt.legend()
plt.title("Peak Detection Results")
plt.show()
```

```python
import numpy as np
from scipy.stats import skew, kurtosis, zscore

# 샘플 데이터 생성
np.random.seed(42)
data = np.random.normal(0, 1, 1000)
data[500:510] += 5  # 이상치 추가

# Modified Z-Score
def modified_zscore(data):
    median = np.median(data)
    mad = np.median(np.abs(data - median))
    return 0.6745 * (data - median) / mad

mz_scores = modified_zscore(data)

# Peak-to-Signal Ratio
psr = (np.max(data) - np.min(data)) / np.mean(data)

# Coefficient of Variation
cv = np.std(data) / np.mean(data)

# Skewness and Kurtosis
data_skewness = skew(data)
data_kurtosis = kurtosis(data)

# Range Ratio
q1, q3 = np.percentile(data, [25, 75])
range_ratio = (q3 - q1) / (np.max(data) - np.min(data))

# Peak-to-Median Ratio
pmr = np.max(data) / np.median(data)

# 결과 출력
print("Modified Z-Score:", mz_scores[:10])  # 일부 출력
print("Peak-to-Signal Ratio (PSR):", psr)
print("Coefficient of Variation (CV):", cv)
print("Skewness:", data_skewness)
print("Kurtosis:", data_kurtosis)
print("Range Ratio:", range_ratio)
print("Peak-to-Median Ratio (PMR):", pmr)
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# 1. 평균과 표준편차 기반 Threshold

def mean_std_threshold(data, k=3):
    mean = np.mean(data)
    std = np.std(data)
    lower = mean - k * std
    upper = mean + k * std
    return lower, upper

# 2. 중앙값과 MAD 기반 Threshold

def mad_threshold(data, k=3):
    median = np.median(data)
    mad = np.median(np.abs(data - median))
    lower = median - k * mad
    upper = median + k * mad
    return lower, upper

# 3. IQR 기반 Threshold

def iqr_threshold(data, k=1.5):
    q1 = np.percentile(data, 25)
    q3 = np.percentile(data, 75)
    iqr = q3 - q1
    lower = q1 - k * iqr
    upper = q3 + k * iqr
    return lower, upper

# 4. Percentile 기반 Threshold

def percentile_threshold(data, p=95):
    lower = np.percentile(data, 100 - p)
    upper = np.percentile(data, p)
    return lower, upper

# 5. 동적 Threshold (Moving Average)

def moving_average_threshold(data, window=10, k=2):
    rolling_mean = pd.Series(data).rolling(window=window).mean()
    rolling_std = pd.Series(data).rolling(window=window).std()
    lower = rolling_mean - k * rolling_std
    upper = rolling_mean + k * rolling_std
    return lower, upper

# 샘플 데이터 생성
np.random.seed(42)
data = np.random.normal(0, 1, 1000)
data[500:510] += 5  # 피크 추가

# Threshold 계산
mean_lower, mean_upper = mean_std_threshold(data, k=3)
mad_lower, mad_upper = mad_threshold(data, k=3)
iqr_lower, iqr_upper = iqr_threshold(data, k=1.5)
percentile_lower, percentile_upper = percentile_threshold(data, p=95)

# 동적 Threshold 계산
moving_lower, moving_upper = moving_average_threshold(data, window=50, k=2)

# 시각화
plt.figure(figsize=(12, 6))
plt.plot(data, label="Data", alpha=0.7)
plt.axhline(mean_upper, color='red', linestyle='--', label="Mean ± 3*Std")
plt.axhline(mean_lower, color='red', linestyle='--')
plt.axhline(mad_upper, color='green', linestyle='--', label="Median ± 3*MAD")
plt.axhline(mad_lower, color='green', linestyle='--')
plt.axhline(iqr_upper, color='blue', linestyle='--', label="IQR-Based Threshold")
plt.axhline(iqr_lower, color='blue', linestyle='--')
plt.axhline(percentile_upper, color='purple', linestyle='--', label="95th Percentile Threshold")
plt.axhline(percentile_lower, color='purple', linestyle='--')
plt.plot(moving_upper, color='orange', linestyle='--', label="Moving Avg + 2*Std")
plt.plot(moving_lower, color='orange', linestyle='--')
plt.legend()
plt.title("Threshold Examples")
plt.show()
```
