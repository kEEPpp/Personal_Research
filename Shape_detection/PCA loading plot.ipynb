```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

# 예제 데이터 생성
np.random.seed(42)
X = np.random.rand(100, 4)
feature_names = ['Var1', 'Var2', 'Var3', 'Var4']

# 표준화
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# PCA 수행
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

# loading 추출
loadings = pca.components_.T  # shape: (n_features, n_components)

# loading plot
plt.figure(figsize=(6, 6))
for i, var in enumerate(feature_names):
    plt.arrow(0, 0, 
              loadings[i, 0], loadings[i, 1],
              head_width=0.05, color='r', alpha=0.5)
    plt.text(loadings[i, 0]*1.15, loadings[i, 1]*1.15, var, 
             color='blue', ha='center', va='center')
    
plt.axhline(0, color='gray', lw=1)
plt.axvline(0, color='gray', lw=1)
plt.xlabel('PC1')
plt.ylabel('PC2')
plt.title('PCA Loading Plot')
plt.grid()
plt.show()
```

```python
loadings = pca.components_.T  # shape (n_features, n_components)
pc1_loadings = loadings[:, 0]  # PC1에 대한 각 변수의 기여도
```

```python
for var, loading_val in sorted(zip(feature_names, pc1_loadings), key=lambda x: abs(x[1]), reverse=True):
    print(f'{var}: {loading_val:.4f}')
```
